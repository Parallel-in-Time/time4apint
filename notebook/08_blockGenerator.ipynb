# BlockIteration generator

Using the formalism of block iterations, one can generate different algorithms using symbolic computations. In particular, we can use the time-multigrid formalism to determine the block iteration for Parareal and many of its extension.

For instance, Parareal consist on a 2-level time-multigrid approach with 1 pre-smoothing step, and the block update formula can be obtained like this :

````python
from blockops.blockIterationGenerator import MultilevelGenerator

MultilevelGenerator(nBlocks=7, nLevels=2, nPreSmooth=1, nPostSmooth=0);
````

Now, if we consider 2 pre-smoothing, then we get Parareal with overlap (or MGRIT with FCF relaxation) :

````python
MultilevelGenerator(nBlocks=7, nLevels=2, nPreSmooth=2, nPostSmooth=0);
````

and this can be continued to get FCFCF[...] relaxation :

````python
MultilevelGenerator(nBlocks=7, nLevels=2, nPreSmooth=3, nPostSmooth=0);
````

Now, more interesting could be to build a block formula for the three-level Parareal approach :

````python
MultilevelGenerator(nBlocks=7, nLevels=3, nPreSmooth=1, nPostSmooth=0);
````

... and even go to 4 levels eventually :

````python
MultilevelGenerator(nBlocks=7, nLevels=4, nPreSmooth=1, nPostSmooth=0);
````

Note that until now, the formula were automatically simplified using those notations :

$$
\begin{align}
F &= \phi_0^{-1}\chi_0 \\
G &= T_0^1 \phi_1^{-1}\chi_1 T_1^0 \\
H &= T_0^1 T_1^2 \phi_2^{-1}\chi_2 T_2^1 T_1^0 \\
K &= T_0^1 T_1^2 T_2^3 \phi_3^{-1}\chi_3 T_3^2 T_2^1 T_1^0 \\
...&
\end{align}
$$

where $0$ is the index used for the finest level, and the increases with the coarsening. $T_{i}^{j}$ is then the transfer operator to go from level $i$ to level $j$.

Considering the hypothesis that if $j<i$, $T_{i}^{j} T_{j}^{i} = I$ (which means that going from a coarse level to the fine level above then back to the coarse level again does not change the solution), we can simplify the formula for 4-level Parareal above as :

```latex
u_{n+1}^{k+1}=
    F*u_{n}^{k}
    +K*u_{n}^{k+1}
    -G*u_{n}^{k}
    +G*F*u_{n-1}^{k}
    -K*F*u_{n-1}^{k}
    -H*G*u_{n-1}^{k}
    +H*G*F*u_{n-2}^{k}
    +K*G*u_{n-1}^{k}
    -K*G*F*u_{n-2}^{k}
```

Now, don't hesitate to play and generate all kind of Parareal-like formula !
